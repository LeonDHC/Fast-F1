```python
import matplotlib.pyplot as plt
from itertools import product
import random

Number_of_Laps = 50
# Track_Length = 4.5  # km
# Track_Temp = 25  # C
# Track_Condition = 1.0  # 1.0 = dry, 0.8 = wet, 0.5 = very wet

# create a random number of laps where SC is active for 5 laps (sc_duration)
def generate_sc_schedule(number_of_laps, sc_probability=0.0125, sc_duration=5):
    sc_laps = []
    for lap in range(1, number_of_laps + 1):
        if random.random() < sc_probability and (not sc_laps or lap > sc_laps[-1] + sc_duration):
            sc_laps.extend(range(lap, min(lap + sc_duration, number_of_laps + 1)))
    return sc_laps

# Generate the safety car schedule for given number of laps
shared_sc_schedule = generate_sc_schedule(Number_of_Laps)
print(f"Safety Car in Laps: {shared_sc_schedule}")

#Hyperparameters
BasePace = {"Soft": 94.5, "Medium": 95.5, "Hard": 97.5}  # Base pace for each tire compound
TireDegradation = {"Soft": 1.5, "Medium": 1.25, "Hard": 1.1}  # Base degradation per lap for each tire
Tires = ["Soft", "Medium", "Hard"]

Sensetivity_Weight = 1.5  # s/kg
Fuel_Start = 45  # kg
Fuel_Burn = 0.75  # kg/lap

StopTime_Delta = 24  # s
StopTime_Delta_SC = 8  # s
StopTime_Delta_VSC = 16  # s

# Simulating the total time it takes under a given tire strategy, pit laps and safety car schedule
def simulate_race(strategy, pit_laps, sc_schedule, plot=False):
    CumulativeLapTime = 0.0
    Fuel_Remaining = Fuel_Start
    CurrentLap = 1
    CurrentTire = strategy[0]  # Start with the first tire in the strategy
    lap_times = []  # Store lap times for plotting
    lap_numbers = []  # Store lap numbers for plotting
    tire_colors = []  # Store tire colors for plotting
    laps_on_tire = 0  # Track laps completed on the current tire
    sc_laps = []  # Track laps where SC is active

    for stint_index, stint_length in enumerate(pit_laps + [Number_of_Laps - sum(pit_laps)]): 
        for lap in range(stint_length):
            if CurrentLap > Number_of_Laps:
                break  # Stop if we've completed all laps

            Fuel_Burnt = Fuel_Start - Fuel_Remaining

            # Check if the current lap is in the predefined SC schedule
            SC_Active = CurrentLap in sc_schedule
            if SC_Active:
                sc_laps.append(CurrentLap)  # Record the lap as under safety car
                CurrentLapTime = (
                    BasePace[CurrentTire]
                    + TireDegradation[CurrentTire] * (1.0 ** laps_on_tire)  # Reduce Tire Deg under SC
                    + BasePace[CurrentTire] * 0.5 * SC_Active # no fuel burnt advantage
                )
            else:
                CurrentLapTime = (
                    BasePace[CurrentTire]
                    + TireDegradation[CurrentTire] * (1.15 ** laps_on_tire)  # Exponential degradation
                    - (Sensetivity_Weight * Fuel_Burnt)
                )

            # If there is a pit stop this lap, add the pit stop time delta
            if lap == stint_length - 1 and stint_index < len(pit_laps):
                # Adjust pit stop time based on SC_Active
                if SC_Active:
                    CurrentLapTime += StopTime_Delta_SC
                else:
                    CurrentLapTime += StopTime_Delta

            CumulativeLapTime += CurrentLapTime
            Fuel_Remaining -= Fuel_Burn

            # Store data for plotting
            lap_times.append(CurrentLapTime)
            lap_numbers.append(CurrentLap)
            if CurrentTire == "Soft":
                tire_colors.append("red")
            elif CurrentTire == "Medium":
                tire_colors.append("yellow")
            elif CurrentTire == "Hard":
                tire_colors.append("white")

            CurrentLap += 1
            laps_on_tire += 1  # Increment laps on the current tire

        # Switch to the next tire after the pit stop
        if stint_index < len(pit_laps):
            CurrentTire = strategy[stint_index + 1]  # Switch to the next tire
            laps_on_tire = 0  # Reset laps on the new tire

    if plot:
        # Plot the lap times with a black background
        plt.figure(figsize=(10, 6), facecolor="black")
        ax = plt.gca()
        ax.set_facecolor("black")

        # Highlight SC active phases with red bars (only if SC laps exist)
        if sc_laps:  # Check if there are any SC-active laps
            start_sc = sc_laps[0]
            for i in range(1, len(sc_laps)):
                if sc_laps[i] != sc_laps[i - 1] + 1:  # Check if the SC phase ends
                    plt.axvspan(start_sc, sc_laps[i - 1] + 1, color="red", alpha=0.3, label="SC Phase")
                    start_sc = sc_laps[i]
            # Add the last SC phase
            plt.axvspan(start_sc, sc_laps[-1] + 1, color="red", alpha=0.3, label="SC Phase")

        # Plot lap times
        for i in range(len(lap_numbers) - 1):
            plt.plot(
                lap_numbers[i:i+2],  # Plot each lap as a segment
                lap_times[i:i+2],
                color=tire_colors[i],
                linewidth=2,
            )

        plt.xlabel("Lap Number", color="white")
        plt.ylabel("Lap Time (s)", color="white")
        plt.title("Lap Time Sim (Fastest Strategy)", color="white")
        plt.grid(color="gray", linestyle="--", linewidth=0.5)
        ax.tick_params(colors="white")  # Set tick colors to white

        # Add legend only if SC phases exist
        if sc_laps:
            plt.legend(loc="upper left", fontsize="small", facecolor="black", edgecolor="white", labelcolor="white")

    plt.show()

    return CumulativeLapTime, sc_laps

# Generate all possible strategies (combinations of tires and pit stop laps)
strategies = []
for num_tires in range(2, 4):  # Allow up to 4 tire sets
    for perm in product(Tires, repeat=num_tires):  # Allow repeated tire usage
        if len(set(perm)) < 2:  # Ensure at least two different types of tires are used
            continue  # Skip strategies with only one type of tire
        for pit_laps in product(range(1, Number_of_Laps), repeat=num_tires - 1):  # Generate pit stop combinations
            if sum(pit_laps) < Number_of_Laps:  # Ensure total pit laps don't exceed total laps
                strategies.append((perm, list(pit_laps)))

# Simulate all strategies and find the fastest one
best_time = float("inf")
best_strategy = None
best_sc_laps = None

for strategy, pit_laps in strategies:
    total_time, sc_laps = simulate_race(strategy, pit_laps, sc_schedule=shared_sc_schedule, plot=False)
    if total_time < best_time:
        best_time = total_time
        best_strategy = (strategy, pit_laps)
        best_sc_laps = sc_laps  # Store SC laps for the fastest strategy

# Output the best strategy and its total time
cumulative_pit_laps = [sum(best_strategy[1][:i+1]) for i in range(len(best_strategy[1]))]  # Calculate cumulative lap numbers
print(f"Best Strategy: {best_strategy[0]} with pit stops at laps {cumulative_pit_laps}")
print(f"Total Time: {best_time:.2f} seconds")

# Plot the fastest strategy
simulate_race(best_strategy[0], best_strategy[1], sc_schedule=shared_sc_schedule, plot=True)
```
