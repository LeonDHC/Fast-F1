```python
session = fastf1.get_session(2025, 2, 'R')
session.load(telemetry=False, weather=False)

fig, ax = plt.subplots(figsize=(8.0, 4.9))

# Create a dictionary to store driver styles and their final positions
driver_styles = {}

for drv in session.drivers:
    drv_laps = session.laps.pick_drivers(drv)

    abb = drv_laps['Driver'].iloc[0]
    style = fastf1.plotting.get_driver_style(identifier=abb,
                                             style=['color', 'linestyle'],
                                             session=session)

    ax.plot(drv_laps['LapNumber'], drv_laps['Position'],
            label=abb, **style)
    
    # Store the final position of the driver
    final_position = drv_laps['Position'].iloc[-1]
    driver_styles[abb] = (final_position, style)

# Sort drivers by their final position
sorted_drivers = sorted(driver_styles.items(), key=lambda x: x[1][0])

# Add the legend in the sorted order
handles, labels = ax.get_legend_handles_labels()
sorted_handles_labels = sorted(zip(handles, labels), key=lambda x: [d[0] for d in sorted_drivers].index(x[1]))
sorted_handles, sorted_labels = zip(*sorted_handles_labels)

ax.legend(sorted_handles, sorted_labels, bbox_to_anchor=(1.0, 1.02))

ax.set_ylim([20.5, 0.5])
ax.set_yticks([1, 5, 10, 15, 20])
ax.set_xlabel('Lap')
ax.set_ylabel('Position')

plt.tight_layout()
plt.show()
```
